# Toon format exploration

The `toon` package can be obtained from [official toon (python)](https://github.com/toon-format/toon-python).

The examples below are obtained from there.

```python
from toon_format import encode, decode

# Simple object
text = encode({"name": "Alice", "age": 30})
print(text)
```

```python
# Tabular array (uniform objects)
text = encode([{"id": 1, "name": "Alice"}, {"id": 2, "name": "Bob"}])
print(text)
```

```python
decoded_text = decode(text)
print(decoded_text)
```

# Token usage comparison

**Assumption:** The token counter is the same as GPT-4o’s.

```python
from pkz.toon.module.format_comparator import FormatComparator
comparator = FormatComparator()
```

## Example 1: Toy example

```python
data = [
    {"id": 1, "name": "Alice", "age": 30, "city": "New York"},
    {"id": 2, "name": "Bob", "age": 25, "city": "San Francisco"},
    {"id": 3, "name": "Charlie", "age": 35, "city": "Los Angeles"},
    {"id": 4, "name": "Diana", "age": 28, "city": "Chicago"},
    {"id": 5, "name": "Eve", "age": 32, "city": "Boston"},
]


comparator.print_comparison(data, include_csv=True, include_xml=True)
```

```python
# json
print(comparator.display_json(data))
```

```python
# json compact
print(comparator.display_json_compact(data))
```

```python
# csv
print(comparator.display_csv(data))
```

```python
# toon
print(comparator.display_toon(data))
```
